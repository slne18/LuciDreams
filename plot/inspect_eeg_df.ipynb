## Extract EEG rows during REM periods

This section loads `rem_episodes.csv`, infers full datetimes for REM start/end from clock-only times, and prints the EEG DataFrame for each REM period.

```python
from datetime import datetime, timedelta
from IPython.display import display

REM_CSV = os.path.join(BASE_DIR, "rem_episodes.csv")
if not os.path.exists(REM_CSV):
    raise FileNotFoundError(f"Missing file: {REM_CSV}")

rem_df = pd.read_csv(REM_CSV)

# Optional: set to a participant id like "Soso" to filter REM extraction.
# Leave as None to keep all rows.
PID_FILTER = "Soso"
if PID_FILTER is not None:
    if "pid" not in rem_df.columns:
        raise ValueError("PID_FILTER is set but rem_episodes.csv has no 'pid' column.")
    rem_df = rem_df[rem_df["pid"].astype(str) == str(PID_FILTER)].copy()
    if rem_df.empty:
        raise ValueError(f"No REM rows found for pid={PID_FILTER} in {REM_CSV}")

# Flexible column detection (supports current and older export variants)
start_candidates = [
    "episode_start_boston",
    "episode_start",
    "rem_start_boston",
    "rem_start",
]
end_candidates = [
    "episode_end_boston",
    "episode_end",
    "rem_end_boston",
    "rem_end",
]

start_col = next((c for c in start_candidates if c in rem_df.columns), None)
end_col = next((c for c in end_candidates if c in rem_df.columns), None)

if start_col is None or end_col is None:
    raise ValueError(
        f"Could not find REM start/end columns in {REM_CSV}. "
        f"Found columns: {list(rem_df.columns)}"
    )

if "TimeStamp" not in df.columns:
    raise ValueError("EEG DataFrame has no TimeStamp column.")

eeg_ts = pd.to_datetime(df["TimeStamp"], errors="coerce").dropna()
if eeg_ts.empty:
    raise ValueError("No valid EEG timestamps found in df['TimeStamp'].")

eeg_start = eeg_ts.min().to_pydatetime()
eeg_end = eeg_ts.max().to_pydatetime()

print(f"REM source file: {REM_CSV}")
print(f"Using start column: {start_col}")
print(f"Using end column  : {end_col}")
print(f"EEG range         : {eeg_start} -> {eeg_end}")


def parse_clock_to_recording_datetime(clock_text: str, recording_start: datetime, recording_end: datetime) -> datetime:
    """Map HH:MM:SS clock text to the correct recording day (start date or next day)."""
    t = datetime.strptime(str(clock_text), "%H:%M:%S").time()
    base_day = recording_start.date()

    c0 = datetime.combine(base_day, t)
    c1 = c0 + timedelta(days=1)

    # Prefer candidates inside recording window.
    in0 = recording_start <= c0 <= recording_end
    in1 = recording_start <= c1 <= recording_end

    if in0 and not in1:
        return c0
    if in1 and not in0:
        return c1

    # If both/none are inside, choose the one closest to recording start.
    return c0 if abs((c0 - recording_start).total_seconds()) <= abs((c1 - recording_start).total_seconds()) else c1


df_ts = df.copy()
df_ts["TimeStamp"] = pd.to_datetime(df_ts["TimeStamp"], errors="coerce")

df_ts = df_ts.dropna(subset=["TimeStamp"]).sort_values("TimeStamp")

rem_slices = []

for i, row in rem_df.iterrows():
    s_txt = row.get(start_col)
    e_txt = row.get(end_col)

    if pd.isna(s_txt) or pd.isna(e_txt):
        continue

    start_dt = parse_clock_to_recording_datetime(str(s_txt), eeg_start, eeg_end)
    end_dt = parse_clock_to_recording_datetime(str(e_txt), eeg_start, eeg_end)

    # Handle midnight crossover inside one REM period, if present.
    if end_dt < start_dt:
        end_dt += timedelta(days=1)

    window_df = df_ts[(df_ts["TimeStamp"] >= start_dt) & (df_ts["TimeStamp"] <= end_dt)].copy()

    rem_label = row.get("rem_episode_index", row.get("episode_index", i))
    print("\n" + "=" * 80)
    print(f"REM episode {rem_label} | {start_dt} -> {end_dt} | rows={len(window_df)}")

    display(window_df)

    rem_slices.append(
        {
            "rem_episode": rem_label,
            "start": start_dt,
            "end": end_dt,
            "rows": len(window_df),
            "df": window_df,
        }
    )

print("\nDone. Use rem_slices[k]['df'] to access a specific REM DataFrame slice.")
```

## Full-night EEG plot (RAW_AF7 and RAW_AF8)

This plot shows the entire overnight recording and marks the last valid sample for each channel to help identify when a channel stopped working.

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

if "TimeStamp" not in df.columns:
    raise ValueError("df must contain a TimeStamp column.")


def pick_col(candidates):
    for c in candidates:
        if c in df.columns:
            return c
    return None


def robust_zscore(series):
    vals = pd.to_numeric(series, errors="coerce").astype(float)
    med = float(np.nanmedian(vals))
    mad = float(np.nanmedian(np.abs(vals - med)))
    scale = 1.4826 * mad
    if not np.isfinite(scale) or scale < 1e-9:
        scale = float(np.nanstd(vals))
    if not np.isfinite(scale) or scale < 1e-9:
        scale = 1.0
    return (vals - med) / scale


def format_elapsed_hm(seconds):
    s = int(max(0, round(seconds)))
    h = s // 3600
    m = (s % 3600) // 60
    return f"{h:02d}:{m:02d}"


af7_col = pick_col(["RAW_AF7", "AF7", "raw_af7"])
af8_col = pick_col(["RAW_AF8", "AF8", "raw_af8"])
if af7_col is None and af8_col is None:
    raise ValueError(f"Could not find AF7/AF8 columns in df. Found: {list(df.columns)}")

plot_df = df.copy()
plot_df["TimeStamp"] = pd.to_datetime(plot_df["TimeStamp"], errors="coerce")
plot_df = plot_df.dropna(subset=["TimeStamp"]).sort_values("TimeStamp")
for col in [af7_col, af8_col]:
    if col is not None:
        plot_df[col] = pd.to_numeric(plot_df[col], errors="coerce")

if plot_df.empty:
    raise ValueError("No valid EEG rows after timestamp parsing.")

# Match plot_eeg_rem.py overview logic: 1-second aggregation + robust normalization.
start_ts = plot_df["TimeStamp"].iloc[0]
end_ts = plot_df["TimeStamp"].iloc[-1]
plot_df["elapsed_sec"] = (plot_df["TimeStamp"] - start_ts).dt.total_seconds()
plot_df["elapsed_bin"] = np.floor(plot_df["elapsed_sec"]).astype(int)

agg = plot_df.groupby("elapsed_bin", as_index=False).agg(
    AF7=(af7_col, "mean") if af7_col is not None else (plot_df.columns[0], "size"),
    AF8=(af8_col, "mean") if af8_col is not None else (plot_df.columns[0], "size"),
)

if af7_col is not None:
    agg["AF7_plot"] = robust_zscore(agg["AF7"]).clip(-6, 6)
if af8_col is not None:
    agg["AF8_plot"] = robust_zscore(agg["AF8"]).clip(-6, 6)

max_points = 40_000
if len(agg) > max_points:
    step = int(np.ceil(len(agg) / max_points))
    agg = agg.iloc[::step].copy()
    print(f"Downsampled aggregated view: every {step}th second-bin ({len(agg)} points shown)")

fig, ax = plt.subplots(figsize=(18, 7))
if af7_col is not None:
    ax.plot(agg["elapsed_bin"], agg["AF7_plot"], color="tab:blue", lw=0.8, alpha=0.9, label=f"{af7_col} (robust z)")
if af8_col is not None:
    ax.plot(agg["elapsed_bin"], agg["AF8_plot"], color="tab:orange", lw=0.8, alpha=0.9, label=f"{af8_col} (robust z)")

# Keep channel stop markers (on elapsed-time axis).
for col, key, color in [(af7_col, "AF7", "tab:blue"), (af8_col, "AF8", "tab:orange")]:
    if col is None:
        continue
    valid = plot_df[["TimeStamp", col]].dropna(subset=[col])
    if valid.empty:
        print(f"{col}: no valid samples")
        continue
    t_stop = valid["TimeStamp"].iloc[-1]
    x_stop = (t_stop - start_ts).total_seconds()
    ax.axvline(x_stop, linestyle="--", linewidth=1.0, alpha=0.75, color=color, label=f"{col} last valid")
    print(f"{col} last valid sample: {t_stop}")

# Elapsed-time ticks every 30 minutes, as in overview plotting style.
total_sec = int(agg["elapsed_bin"].max()) if not agg.empty else 0
tick_step = 30 * 60
ticks = np.arange(0, max(total_sec + tick_step, tick_step), tick_step)
ax.set_xticks(ticks)
ax.set_xticklabels([format_elapsed_hm(t) for t in ticks])

ax.set_title(f"Full-night EEG overview ({start_ts:%Y-%m-%d %H:%M:%S} -> {end_ts:%Y-%m-%d %H:%M:%S})")
ax.set_xlabel("Elapsed from start (HH:MM)")
ax.set_ylabel("Normalized EEG (robust z)")
ax.grid(alpha=0.25)
ax.legend(loc="upper right", fontsize=8)
plt.tight_layout()
plt.show()
```
